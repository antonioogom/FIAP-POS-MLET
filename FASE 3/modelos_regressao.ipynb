# Modelos de Regressão

Algumas literaturas apontem que a classificação pode ser um problema de regressão, devido a sua similaridade.
O que difere é a coluna alvo.

Exemplo: predição de valores estimados de casas;
Uma casa com 2 quartos, 1 coz, 3 banheiros pode valer 30mil
Uma casa com mais quartos pode ter preço maior, dependendo de bairro ou outros fatores
