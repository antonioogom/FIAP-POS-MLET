# Identificar Outliers
(valores atipicos ou muito discrepantes. Falhou o sensor, ou comportamento anormal)
Eles podem distorcer o modelo e levar previsões fora do esperado, tomar decições erradas etc.
Pode ser ou não uma tomada de decisão do time tirar os outliers, mas identificar é importante.

```python
# IQR Divisão interquartilica. Dividimos em quartis, ou seja, em 4 partes. O primeiro quartil é o valor que separa os 25% menores dos 75% maiores. O segundo quartil é o valor que separa os 50% menores dos 50% maiores. O terceiro quartil é o valor que separa os 75% menores dos 25% maiores. O IQR é a diferença entre o terceiro quartil e o primeiro quartil. Valores que estão abaixo do primeiro quartil - 1.5 * IQR ou acima do terceiro quartil + 1.5 * IQR são considerados outliers.
# Depois obsevamos o que esta dentro do intervalo de quartil.
```

```python
dados = np.array([1, 2, 3, 4, 5, 100])
```

```python
dados
```

```python
q1 = np.percentile(dados, 25)
q3 = np.percentile(dados, 75)
iqr = q3 - q1
```

```python
iqr
```

```python
# Calculo do limite inferior e superior para identificar outliers
# Nos livros de estatistica adota 1,5 para identificar outliers, ou seja, valores que estão abaixo do primeiro quartil - 1.5 * IQR ou acima do terceiro quartil + 1.5 * IQR são considerados outliers.
limite_inf = q1 - 1.5 * iqr
limite_sup = q3 + 1.5 * iqr
```

```python
limite_inf, limite_sup # Os valores estão considerados dentro desse intervalo. O que esta fora desse intervalo é considerado outlier. O 100 é considerado outlier, pois esta acima do limite superior.
```

```python
outliers = np.where((dados < limite_inf) | (dados > limite_sup))[0]
```

```python
outliers
```

```python
dados[outliers[0]] # mostra o outlier, ou seja, o valor 100.
```

# Calculo das métricas de acurácia, precisão, revocação e f1-score

Exemplo relacionado a testes de COVID-19
Imagine um teste de COVID em um população de 1000 pessoas. Dessas 1000, 200 estão realmente infectadas e as outras 800 saudáveis.

Definições

Verdadeiro Positivo (VP):
    Quando a pessoa está positiva e o teste deu positivo

Verdadeiro Negativo (VN):
    Quando o resultado é negativo e a pessoa realmente não está infectada

Falso positivo (FP):
    Quando o teste é positivo mas a pessoa não está infectada

Falso negativo (FN):
    Quando o teste é negativo, mas a pessoa está infectada.

Infectadas = 200
Não infectadas = 800

VP = 150

VN = 700

FP = 50

FN = 50

```python
total = 1000
vp = 150
vn = 750
fp = 50
fn = 50
```

# Acurácia (chamada de taxa de acerto, para avalisar desempenho do modelo)

```python
acuracia = (vp + vn) / total
acuracia
```

# Precisão
Relevancia das previsões positivas
Chamada tambem, valor preditivo positivo
Proporção de classificação positivas pelo modelo (acertadas)

- Minimizar custos de falso positivo, identificar crimes, golpes etc. Usado quando o modelo precisa ser preciso (saude por exemplo)

```python
precisao = vp / (vp + fp)
precisao
```

# Revocação ou Recall
Relacionado a proporção de resultados positivos reais que foram identificados pelo modelo.

- Minimizar custos de falso positivo, identificar crimes, golpes etc. Usado quando o modelo precisa ser preciso (saude por exemplo)

```python
revocacao = vp / (vp + fn)
revocacao
```

# F1-Score

Técnica que junta os dois (recall e precisao), vc combina e olha para apenas uma, tendo uma informação mais equilibrada

```python
f1_score = 2 * (precisao * revocacao) / (precisao + revocacao)
f1_score
# 1 o modelo perfeito, 0 o pior cenário
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
y_real = [1, 0, 1, 0]
y_previsto = [0, 1, 1, 1]
```

```python
print(f"Acurácia: {accuracy_score(y_real, y_previsto)}")
print(f"Precisão: {precision_score(y_real, y_previsto)}")
print(f"Revocação: {recall_score(y_real, y_previsto)}")
print(f"F1-Score: {f1_score(y_real, y_previsto)}")
```

```python
from sklearn.metrics import classification_report
```

```python
target_names = ['Classe 0', 'Classe 1']
print(classification_report(y_real, y_previsto, target_names=target_names))
```
